```python
import torch

# Basit bir tensor oluştur (normal bir Python listesinden)
data = [1, 2, 3, 4]
x = torch.tensor(data)

print(x)
print("Tipi:", type(x))
print("Boyutu (shape):", x.shape)
```

```python
# requires_grad=True: PyTorch'a "bu tensor üzerinde yapılan işlemleri takip et, 
# çünkü ileride türevini alacağım" diyoruz
x = torch.tensor(3.0, requires_grad=True)

# Basit bir matematiksel işlem: y = x^2
y = x ** 2

# y'nin x'e göre türevini otomatik hesapla
y.backward()

print("x:", x.item())
print("y:", y.item())
print("dy/dx (x'deki gradient):", x.grad.item())
```

```python
# Öğrenilecek "gizli" ilişki: y = 2x (ama modele bunu söylemiyoruz)
x_train = torch.tensor([1.0, 2.0, 3.0, 4.0])
y_train = torch.tensor([2.0, 4.0, 6.0, 8.0])

# Modelin öğreneceği tek parametre: w (ağırlık). Rastgele bir değerle başlıyoruz.
w = torch.tensor(0.0, requires_grad=True)

learning_rate = 0.01

for epoch in range(50):
    # 1. Forward pass: mevcut w ile tahmin yap
    y_pred = w * x_train

    # 2. Loss hesapla (MSE: tahmin ile gerçek arasındaki ortalama kare hata)
    loss = ((y_pred - y_train) ** 2).mean()

    # 3. Backward pass: loss'un w'ye göre gradyanını hesapla
    loss.backward()

    # 4. w'yi güncelle (gradyanın tersi yönünde küçük bir adım at)
    with torch.no_grad():
        w -= learning_rate * w.grad
        w.grad.zero_()  # bir sonraki epoch için gradyanı sıfırla

    if epoch % 10 == 0:
        print(f"Epoch {epoch}: w = {w.item():.4f}, loss = {loss.item():.4f}")

print("\nSon öğrenilen w değeri:", w.item(), "(gerçek değer: 2.0 olmalı)")
```

```python
import torch.nn as nn

# Aynı veri, aynı problem: y = 2x ilişkisini öğrenmeye çalışıyoruz
x_train = torch.tensor([[1.0], [2.0], [3.0], [4.0]])  # dikkat: artık 2 boyutlu
y_train = torch.tensor([[2.0], [4.0], [6.0], [8.0]])

# nn.Linear(1, 1): 1 girdi al, 1 çıktı üret — içinde otomatik olarak
# bir "w" (ağırlık) ve bir "b" (bias) parametresi barındırır
model = nn.Linear(1, 1)

# Kayıp fonksiyonu: MSE'yi elle yazmak yerine hazır kullanıyoruz
loss_fn = nn.MSELoss()

# Optimizer: "w -= learning_rate * w.grad" satırını bizim yerimize o yapacak
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)

for epoch in range(500):
    y_pred = model(x_train)              # forward pass
    loss = loss_fn(y_pred, y_train)      # loss hesapla

    optimizer.zero_grad()                # gradyanları sıfırla (w.grad.zero_() yerine)
    loss.backward()                      # gradyanı hesapla
    optimizer.step()                     # parametreleri güncelle (w -= lr*grad yerine)

    if epoch % 10 == 0:
        print(f"Epoch {epoch}: loss = {loss.item():.4f}")

# Öğrenilen ağırlığı göster
learned_w = model.weight.item()
learned_b = model.bias.item()
print(f"\nÖğrenilen w: {learned_w:.4f} (gerçek: 2.0)")
print(f"Öğrenilen b: {learned_b:.4f} (gerçek: 0.0)")
```
